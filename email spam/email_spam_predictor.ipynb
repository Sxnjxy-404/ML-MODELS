```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score,classification_report
```

```python
df=pd.read_csv("data/email_spam.csv")
```

```python
df=pd.DataFrame(df)
```

```python
df
```

```python
vectorizer = CountVectorizer()
X= vectorizer.fit_transform(df["text"]) #Convert text to numbers
y = df["label"]
X_train,X_test, y_train, y_test= train_test_split(X,y,test_size=0.3, random_state=42)
model = MultinomialNB()
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
```

```python
print("Accuracy:", accuracy_score(y_test,y_pred))
print("Report in", classification_report(y_test,y_pred))
```

```python
new_messages=["Congratulations! You've won a free iPhone. Click here to claim."]
new_features=vectorizer.transform(new_messages)
predictions=model.predict(new_features)
```

```python
predictions
```

```python
for msg, label in zip(new_messages, predictions):
    print(f"'{msg}'-> {'Spam' if label == 1 else'Not Spam'}")
```
